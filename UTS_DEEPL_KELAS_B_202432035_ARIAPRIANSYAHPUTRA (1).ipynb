# BAB 2. IMPLEMENTASI PRA-PEMROSESAN DATA

## 2.1 Import Library

Library yang digunakan meliputi NumPy dan Pandas untuk pengolahan data, Matplotlib untuk visualisasi, PyTorch untuk membangun model LSTM, serta Scikit-learn untuk pembagian data dan evaluasi model.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432035_AriApriansyahPutra
import os, re, random, zipfile, glob
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report, confusion_matrix,
    ConfusionMatrixDisplay, accuracy_score
)
from sklearn.utils.class_weight import compute_class_weight

# Menetapkan seed agar hasil pembagian data dan inisialisasi lebih konsisten.
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

# Memilih GPU jika tersedia, jika tidak menggunakan CPU.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Perangkat yang digunakan:", device)
```

**Penjelasan kode:** Import library menyediakan fungsi untuk membaca data, membersihkan teks, membangun model, melatih model, dan mengevaluasi hasil. Seed ditetapkan agar proses acak lebih konsisten. Program memilih GPU jika tersedia.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432035_AriApriansyahPutra
# 1. Mencari file CSV pada folder kerja Colab.
csv_files = glob.glob("*.csv")

# Jika CSV belum ditemukan tetapi file ZIP tersedia, ekstrak ZIP.
if not csv_files and glob.glob("*.zip"):
    with zipfile.ZipFile(glob.glob("*.zip")[0]) as zip_data:
        zip_data.extractall(".")
    csv_files = glob.glob("*.csv")

if not csv_files:
    raise FileNotFoundError(
        "File CSV belum ditemukan. Upload GojekAppReviewV4.0.0-V4.9.3_Cleaned.csv melalui panel Files Colab."
    )

# Membaca kolom teks dan rating yang diperlukan.
df = pd.read_csv(csv_files[0])[["content", "score"]].copy()
df["content"] = df["content"].fillna("").astype(str)
df["score"] = pd.to_numeric(df["score"], errors="coerce")

# Menghapus baris tanpa rating dan ulasan kosong.
df = df.dropna(subset=["score"])
df = df[df["content"].str.strip().ne("")]
df["score"] = df["score"].astype(int)

# Mengubah rating menjadi label: 0 negatif, 1 netral, 2 positif.
df["label"] = df["score"].map(
    lambda score: 0 if score <= 2 else (1 if score == 3 else 2)
)

# Membersihkan teks: lowercase, URL, karakter non-huruf, dan spasi berlebih.
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-zA-ZÀ-ÿ0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_text"] = df["content"].apply(clean_text)
df = df[df["clean_text"].str.len() > 0].reset_index(drop=True)

# Membatasi jumlah sampel maksimum 60.000 dengan menjaga proporsi kelas.
if len(df) > 60000:
    df, _ = train_test_split(
        df, train_size=60000, random_state=SEED,
        stratify=df["label"]
    )
df = df.reset_index(drop=True)

label_names = {0: "Negatif", 1: "Netral", 2: "Positif"}
print("Jumlah data setelah preprocessing dan sampling:", len(df))
print("\nDistribusi label:")
print(df["label"].map(label_names).value_counts())
display(df.head())
```

**Penjelasan kode:** Program membaca file CSV, memilih kolom `content` dan `score`, menghapus data kosong, lalu mengubah rating menjadi tiga label sentimen. Fungsi `clean_text` melakukan case folding, penghapusan URL dan karakter yang tidak diperlukan, serta merapikan spasi. Jika jumlah data melebihi 60.000, program mengambil sampel secara stratified agar proporsi kelas tetap terjaga. Batas sampling ini digunakan untuk menjaga waktu pelatihan tetap terjangkau.

## 2.2 Pembagian Data, Tokenisasi, dan Padding

Data dibagi menjadi training 72%, validation 8%, dan testing 20%. Vocabulary dibangun hanya dari data training untuk menghindari kebocoran informasi dari data validasi dan pengujian. Setiap kata diubah menjadi indeks angka, kemudian urutan disamakan menjadi 100 token menggunakan padding.

```python
#202432035_AriApriansyahPutra
# Membagi data menjadi training 80% dan testing 20%.
train_texts, test_texts, y_train, y_test = train_test_split(
    df["clean_text"].tolist(),
    df["label"].to_numpy(),
    test_size=0.20,
    random_state=SEED,
    stratify=df["label"]
)

# Membagi 10% dari data training awal menjadi validation.
train_texts, val_texts, y_train, y_val = train_test_split(
    train_texts, y_train,
    test_size=0.10,
    random_state=SEED,
    stratify=y_train
)

# Menentukan ukuran vocabulary dan panjang maksimum teks.
MAX_VOCAB = 20000
MAX_LEN = 100
PAD, UNK = 0, 1

# Membentuk vocabulary hanya dari teks training.
word_counts = Counter(
    token for text in train_texts for token in text.split()
)
vocab = {"<PAD>": PAD, "<UNK>": UNK}
for word, _ in word_counts.most_common(MAX_VOCAB - 2):
    vocab[word] = len(vocab)

# Mengubah teks menjadi indeks dan melakukan padding/truncation.
def encode_text(text):
    token_ids = [vocab.get(token, UNK) for token in text.split()[:MAX_LEN]]
    return token_ids + [PAD] * (MAX_LEN - len(token_ids))

def make_tensor(texts):
    return torch.tensor(
        [encode_text(text) for text in texts],
        dtype=torch.long
    )

X_train = make_tensor(train_texts)
X_val = make_tensor(val_texts)
X_test = make_tensor(test_texts)

y_train_tensor = torch.tensor(y_train, dtype=torch.long)
y_val_tensor = torch.tensor(y_val, dtype=torch.long)
y_test_tensor = torch.tensor(y_test, dtype=torch.long)

# Membuat DataLoader agar data diproses dalam batch.
BATCH_SIZE = 128
train_loader = DataLoader(
    TensorDataset(X_train, y_train_tensor),
    batch_size=BATCH_SIZE, shuffle=True
)
val_loader = DataLoader(
    TensorDataset(X_val, y_val_tensor),
    batch_size=BATCH_SIZE
)
test_loader = DataLoader(
    TensorDataset(X_test, y_test_tensor),
    batch_size=BATCH_SIZE
)

print("Ukuran data training:", X_train.shape)
print("Ukuran data validation:", X_val.shape)
print("Ukuran data testing:", X_test.shape)
print("Jumlah vocabulary:", len(vocab))
```

**Penjelasan kode:** `train_test_split` membagi data secara stratified. Vocabulary dibuat dari data training saja. Token yang tidak terdapat dalam vocabulary diberi indeks `<UNK>`, sedangkan padding menggunakan indeks `<PAD>`. Selanjutnya, data teks dan label diubah menjadi tensor PyTorch dan dimasukkan ke DataLoader agar dapat diproses per batch.

# BAB 3. PEMODELAN ARSITEKTUR JARINGAN LSTM

## 3.1 Arsitektur Model

Model terdiri dari embedding layer, LSTM, dropout, dan fully connected layer. Embedding menghasilkan vektor 128 dimensi untuk setiap token. LSTM menggunakan hidden size 64. Dropout sebesar 0,3 diterapkan sebelum layer klasifikasi. Output akhir memiliki tiga kelas sentimen.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432035_AriApriansyahPutra
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim=128,
                 hidden_size=64, num_classes=3):
        super(SequenceModel, self).__init__()

        # Mengubah indeks token menjadi vektor embedding.
        self.embedding = nn.Embedding(
            vocab_size, embedding_dim, padding_idx=PAD
        )

        # Memproses urutan vektor menggunakan LSTM.
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            batch_first=True
        )

        # Mengurangi risiko overfitting.
        self.dropout = nn.Dropout(0.3)

        # Menghasilkan skor untuk tiga kelas sentimen.
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        embedded = self.embedding(x)
        _, (hidden_state, _) = self.lstm(embedded)
        last_hidden = hidden_state[-1]
        output = self.fc(self.dropout(last_hidden))
        return output

model = SequenceModel(vocab_size=len(vocab)).to(device)
print(model)
```

**Penjelasan kode:** Kelas `SequenceModel` mendefinisikan jaringan LSTM. Fungsi `forward` menjalankan input melalui embedding, LSTM, dropout, dan fully connected layer. Output berupa tiga skor yang digunakan untuk menentukan kelas sentimen.

# BAB 4. HYPERPARAMETER, PELATIHAN, DAN EVALUASI

## 4.1 Konfigurasi Hyperparameter dan Pelatihan

Pelatihan menggunakan optimizer Adam dengan learning rate 0,001 dan delapan epoch. Fungsi Cross Entropy Loss digunakan untuk klasifikasi multikelas. Bobot kelas dihitung dari data training untuk membantu menangani ketidakseimbangan kelas.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432035_AriApriansyahPutra
EPOCHS = 8
LEARNING_RATE = 0.001

# Menghitung bobot kelas berdasarkan distribusi data training.
class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.array([0, 1, 2]),
    y=y_train
)
class_weights = torch.tensor(
    class_weights, dtype=torch.float32
).to(device)

# Menentukan loss function dan optimizer.
criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = torch.optim.Adam(
    model.parameters(), lr=LEARNING_RATE
)

# Fungsi untuk menjalankan satu epoch training atau validation.
def run_epoch(loader, training=False):
    model.train(training)
    total_loss = 0.0
    total_correct = 0
    total_data = 0

    context = torch.enable_grad() if training else torch.no_grad()
    with context:
        for batch_x, batch_y in loader:
            batch_x = batch_x.to(device)
            batch_y = batch_y.to(device)

            if training:
                optimizer.zero_grad()

            logits = model(batch_x)
            loss = criterion(logits, batch_y)

            if training:
                loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()

            total_loss += loss.item() * len(batch_y)
            total_correct += (logits.argmax(1) == batch_y).sum().item()
            total_data += len(batch_y)

    average_loss = total_loss / total_data
    accuracy = total_correct / total_data
    return average_loss, accuracy

# Menyimpan riwayat loss dan accuracy setiap epoch.
history = {
    "train_loss": [], "val_loss": [],
    "train_acc": [], "val_acc": []
}

# Training loop.
for epoch in range(EPOCHS):
    train_loss, train_acc = run_epoch(train_loader, training=True)
    val_loss, val_acc = run_epoch(val_loader, training=False)

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)

    print(
        f"Epoch {epoch + 1}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Validation Loss: {val_loss:.4f} | "
        f"Train Accuracy: {train_acc:.4f} | "
        f"Validation Accuracy: {val_acc:.4f}"
    )
```

**Penjelasan kode:** Program menetapkan hyperparameter, menghitung bobot kelas, dan membuat optimizer Adam. Fungsi `run_epoch` menjalankan proses training atau validasi. Ketika training, gradien dihitung melalui backpropagation dan bobot diperbarui oleh optimizer. Riwayat loss dan accuracy disimpan untuk divisualisasikan.

## 4.2 Visualisasi Loss dan Accuracy

Grafik berikut memperlihatkan perubahan training loss, validation loss, training accuracy, dan validation accuracy selama delapan epoch.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>

```python
#202432035_AriApriansyahPutra
epochs_range = range(1, EPOCHS + 1)

# Grafik loss dan accuracy.
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(epochs_range, history["train_loss"], marker="o", label="Training Loss")
plt.plot(epochs_range, history["val_loss"], marker="o", label="Validation Loss")
plt.title("Training dan Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(epochs_range, history["train_acc"], marker="o", label="Training Accuracy")
plt.plot(epochs_range, history["val_acc"], marker="o", label="Validation Accuracy")
plt.title("Training dan Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 4.3 Evaluasi Model pada Data Testing

Model diuji menggunakan data testing yang tidak digunakan selama proses pembelajaran. Evaluasi meliputi accuracy, precision, recall, F1-score, dan confusion matrix.

```python
#202432035_AriApriansyahPutra
model.eval()
predictions = []
actual_labels = []

with torch.no_grad():
    for batch_x, batch_y in test_loader:
        batch_x = batch_x.to(device)
        logits = model(batch_x)
        batch_predictions = logits.argmax(dim=1).cpu().numpy()

        predictions.extend(batch_predictions)
        actual_labels.extend(batch_y.numpy())

test_accuracy = accuracy_score(actual_labels, predictions)
print(f"Test Accuracy: {test_accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(
    actual_labels,
    predictions,
    labels=[0, 1, 2],
    target_names=["Negatif", "Netral", "Positif"],
    zero_division=0
))

# Confusion matrix.
cm = confusion_matrix(actual_labels, predictions, labels=[0, 1, 2])
display_matrix = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negatif", "Netral", "Positif"]
)
display_matrix.plot(cmap="Blues", values_format="d")
plt.title("Confusion Matrix Data Testing")
plt.show()

# Distribusi prediksi pada data testing.
predicted_counts = pd.Series(predictions).map(
    {0: "Negatif", 1: "Netral", 2: "Positif"}
).value_counts().reindex(["Negatif", "Netral", "Positif"], fill_value=0)

predicted_counts.plot(kind="bar", figsize=(7, 4))
plt.title("Distribusi Prediksi Sentimen pada Data Testing")
plt.xlabel("Kategori Sentimen")
plt.ylabel("Jumlah Prediksi")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Penjelasan kode:** Model diatur ke mode evaluasi dan prediksi dilakukan tanpa menghitung gradien. `accuracy_score` menghitung akurasi keseluruhan. `classification_report` menampilkan precision, recall, dan F1-score per kelas. Confusion matrix memperlihatkan jumlah prediksi benar dan salah, sedangkan grafik batang menunjukkan distribusi kelas hasil prediksi.

## 4.4 Contoh Prediksi Sentimen

Fungsi berikut dapat digunakan untuk menguji kalimat ulasan baru. Hasil prediksi menunjukkan kelas sentimen dan probabilitas masing-masing kelas.

```python
#202432035_AriApriansyahPutra
def predict_sentiment(text):
    cleaned = clean_text(text)
    encoded = encode_text(cleaned)
    input_tensor = torch.tensor([encoded], dtype=torch.long).to(device)

    model.eval()
    with torch.no_grad():
        probabilities = torch.softmax(model(input_tensor), dim=1)[0]

    class_names = ["Negatif", "Netral", "Positif"]
    predicted_index = int(probabilities.argmax().item())
    probability_values = probabilities.cpu().numpy()

    print("Ulasan:", text)
    print("Prediksi sentimen:", class_names[predicted_index])
    print("Probabilitas:")
    for name, probability in zip(class_names, probability_values):
        print(f"- {name}: {probability:.4f}")

predict_sentiment("Aplikasi sangat membantu dan mudah digunakan")
```
